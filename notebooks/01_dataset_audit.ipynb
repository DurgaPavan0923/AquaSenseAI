```python
import os
import sys
from pathlib import Path

PROJECT_ROOT = Path("..").resolve() if Path("..").resolve().name == "AquaSense AI" else Path(".").resolve()
os.chdir(PROJECT_ROOT)
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

import pandas as pd

from src.aquasense.data import FEATURE_COLUMNS, TARGET_COLUMN, audit_dataset_summary, load_raw_dataset

df = load_raw_dataset("data/raw/water_quality.csv")
summary = audit_dataset_summary(df)
print(f"Total Audited Records: {summary['total_samples']}")
print(f"Target Column: {summary['target_column']}")
print(f"Class Distribution: {summary['class_distribution']}")
print(f"Missing Values Count: {summary['missing_values_count']}")
```

```python
# Target class distribution table
class_counts = df[TARGET_COLUMN].value_counts().rename(index={0: "Good (0)", 1: "Moderate (1)", 2: "Polluted (2)"})
class_pct = (df[TARGET_COLUMN].value_counts(normalize=True) * 100).rename(index={0: "Good (0)", 1: "Moderate (1)", 2: "Polluted (2)"})
pd.DataFrame({"Sample Count": class_counts, "Percentage (%)": class_pct.round(2)})
```
